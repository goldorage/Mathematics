[^1]: Un **graphe complet** est un graphe où tous les sommets sont relié deux à deux par une arrête.

## Intuitions

- Une somme $\sum^{g(n)}$ peut toujours se réduire en une expression algébrique $f(n)$, avec $n \in Z^+$.
- Deux sommes $\sum_{i=a}^{g(n)} f_1(n)$ et $\sum_{i=a}^{g(n)} f_2(n)$ de même expression $g(n)$ et $a$ peuvent se simplifier en $\sum_{i=a}^{g(n)} F(n)$. *Quelles relations entre $f_1()$ $f_2()$ et $F()$ ?*
- Le nombre de triangle rectangle qui est possible de tracer dans une tel grille est directement lié au nombre de diagonale qu'il est possible de tracer ; exactement 2 fois plus de triangle que de diagonale.

## Résolution

Intuitivement, on voit bien que certaines de ces problématiques relèves de la théorie de graphes (ex: nb arrêtes totale) alors que d'autre sont purement géométrico-algébrique *(pas sûr du mot)*.


La résolution est purement mathématique et Python permettra la résolution des calculs pour différentes taille de grille et certains visuels.

La résolution de cahque problématique revient à trouver sa fonction mathématique associée.

### Combien la grille comporte-t-elle de carrés (de taille 1) ?

Cela revient à trouver la taille de l'ensemble $V$.

Formule :
$$ n^2 $$

```python
def nbCarrePetit(n:int) -> int:
    return pow(n, 2)
```

### Combien la grille a-t-elle de sommets ?

Formule :
$$ (n+1)^2 $$

Lorsque $n=0$ le nombre de sommet est de $1$.

```python
def nbSommet(n:int) -> int:
    return pow((n+1), 2)
```

### Combien la grille à-t-elle d'arrêtes ?

Cela revient à trouver la taille de l'ensemble $E$.

Question simple de la théorie des graphes ; on sait que comme le graphe est complet, chaque sommet est relié à chaque autres sommets.
Donc pour un graphe de $x$ sommets, un sommet aura $x-1$ arrêtes vers les $x-1$ autres sommets.
Il faut quand même faire attention au fait que lorsqu'une arrête est entre le sommet_A et le sommet_B elle est donc aussi entre sommet_B - sommet_A. Donc le nombre d'arrêtes à compter (si on compte sommet par sommet) et à diminuer de 1 pour chaque sommet déjà compter.

$$\sum_{i=0}^{(n+1)^2 -1} i$$
, avec $(n+1)^2$ le nombre de sommets dans la grille.

Aussi sous forme algébrique :
$$\frac{(n+1)^2((n+1)^2 -1)}{2}$$
On a $(n+1)^2((n+1)^2 -1)$ et pas $(n+1)^2 (n+1)^2$ car, comme dit, chaque arrête entre le sommet_A et le sommet_B elle est aussi entre B et A. Ce qui fait que lorsqu'on voudra compter les arrêtes du dernier sommet, on aura en faite déjà compté toutes ses arrêtes (en comptant c'elles des "sommet_B" respectifs). De même, en moyenne, chaque sommet a $\frac{(n+1)^2}{2}$.

```python
def nbArrete(n:int) -> int:
    return int((pow((n+1), 2) * (pow((n+1), 2)-1)) / (2))
```

### Combien la grille à-t-elle d'arrêtes droites (de taille 1) ?

On commence à $n$ où on a $x$ arrêtes. A chaque incrémentation de $n$, on ajoute une ranger de carrés (sur les deux axes) sur les côtés de la grille existante plus un carré dans l'angle ($n+1$, $n+1$). On ajoute toujours un nombre impair de carrés : $2n-1$. Les carrés ajoutés partagent tous deux côtés avec d'autres carrés (existants pour ce ajoutés sur les côtés de la grille, et pas déjà existant pour le carré de l'angle ($n+1$, $n+1$)).

Cela revient à ajouter la moitié de 4 arrêtes par carrés ajoutés : $\frac{4 (2n-1)}{2}$, donc 2 arrêtes par carrés ajoutés : $2 (2n-1)$.

Le nombre d'arrêtes droites de taille 1 selon $n$ est alors la somme des arrêtes ajoutés à la grille au cours des itérations de 0 à $n$ :

$$\sum_{0}^{n} 4n$$
Aussi sous forme algébrique :
$$
\frac{4n(n+1)}{2}
= 2n(n+1)
$$

```python
def nbArreteDroiteCourte(n:int) -> int:
    return 2*n*(n+1)
```

### Combien la grille à-t-elle d'arrêtes droites (de taille > 1) ?

Trouver le nombre total d'arrêtes droites (de taille > 1) de la grille revient à trouver le nombre d'arrêtes droites (de taille > 1) sur une ligne de taille $n$, puisqu'il est facile de déterminer le nombre de lignes dans une grille.

On peut identifier deux composantes distinctes :
1. la somme des $n-1$ premiers entiers qui correspond au nombre d'arrêtes de taille $> 1$ sur une ligne de taille $n$ : $\sum_{i=0}^{n-1} i$
2. le nombre de lignes dans une grille, avec :
    - le mutiplicateur qui correspond au nombre d'axe de la grille : 2
    - l'expression du nombre de ligne par axe (une ligne est perpendiculaire à l'axe choisit) : $n+1$

Formule :

$$2(n+1) \sum_{i=0}^{n-1} i$$

On explique la somme $\sum_{i=0}^{n-1} i$ par le fait que sur une ligne de taille $n$, on a $n+1$ sommet, et lorsqu'on ajoute un sommet dans un graphe complet (ici notre ligne), on ajouter $n-1$ arrêtes au graphe. Or, dans notre cas on enlève une arrête ; qui est c'elle de taille 1, ce qui fait qu'on ajoute $n-2$ arrêtes. En résumé : la taille $n$, qui a $n+1$ sommets,  ajoute à chaque incrémentations $n-2$ arrêtes (de taille $> 1$). On a alors par l'incrémentation de $n$, un ajout $n-1$ arrêtes, d'où la somme de 0 à $n-1$ pour trouver le nombre total de telles arrêtes sur une ligne.

Aussi sous forme algébrique :

$$
2(n+1) \frac{n(n-1)}{2}
= n^3 -n
$$
, avec $\sum_{i=0}^{n-1} i = \frac{n(n-1)}{2}$.

```python
def nbArreteDroiteLongue(n:int) -> int:
    return int(pow(n, 3)-n)
```

### Combien la grille à-t-elle d'arrêtes diagonales (de taille $\sqrt{2}$) ?

Une arrête diagonale de taille $\sqrt{2}$ correspond à l'hypoténuse d'un triangle rectangle (carré) de côté 1, ce qui est exactement le cas dans la grille ; où chaque carré a 2 diagonales de taille $\sqrt{2}$.

Formule :

$$ 2n^2 $$
, avec $n^2$ le nombre de carrés dans la grille.

```python
def nbArreteDiagonaleCourte(n:int) -> int:
    return int(2*pow(n, 2))
```

### Combien la grille à-t-elle d'arrêtes diagonales (de taille > $\sqrt{2}$) ?

Somme difficile à trouver, car une diagonale de taille > $\sqrt{2}$ est la diagonale d'un rectangle, ce qui revient à trouver tous les rectangles de taille ; $a*b$ dans la grille
, où $a \neq b$ et $a \in (1, 2, .. , n-1)$ et $b \in (2, 3, .. , n)$.

Néanmoins, comme un rectangle à nécessairement 2 diagonales, on sait que l'expression sera de la forme :

$$2 \sum^{g(n)} ?$$
, avec $?$ le nombre de rectangle dans la grille (expression qui dépent bien de $n$).

On peut trouver l'expression ($f(n)$) algébrique d'un autre manière que d'en partant de la somme.
On sait que toutes les arrêtes qui ne sont pas :
- $a(n)$, droites de taille 1;
- $b(n)$, droites de taille > 1;
- $c(n)$, diagonales de taille $\sqrt{2}$
sont le type d'arrêtes qu'on essaye de compter.

Alors :

$$ f(n) = d(n) - (a(n)+b(n)+c(n))$$
, avec $d(n)$ l'expression du nombre totale d'arrêtes.

$$
f(n) =
\frac{(n+1)^2((n+1)^2 -1)}{2} -
(2n(n+1) + (n^3 -n) + 2n^2)
$$

Après développement et réduction :
$$
f(n) = 
\frac{n^4 + 2n^3 - 3n^2}{2}
$$

Ici on peut déduire que $n^4 + 2n^3 - 3n^2$ est un multiple de 4 car on a trouvé que le résultat ($f(n)$) doit être un multiple de 2, et cette expression est divisé par 2.

$$ fn(n)\mod 2 = 0 $$
$$\Leftrightarrow \frac{n^4 + 2n^3 - 3n^2}{2} \mod 2 = 0$$
$$\Leftrightarrow n^4 + 2n^3 - 3n^2 \mod 4 = 0$$


*Cela ne donne surement pas de renseignement sur la formule dans la somme.*


#### Combien la grille a t'elle de rectangles

*Note : Surement un cas particulier de combinaison $(^b_a)$.*

$$ \text{à déterminer} $$

```python
def nbArreteDiagonaleLongue(n:int) -> int:
    return int((pow(n, 4) + 2*pow(n, 3) - 3*pow(n, 2)) / 2)
```
